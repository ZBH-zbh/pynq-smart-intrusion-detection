```python
from pynq import Overlay
import time

ol = Overlay("base.bit")
ol.download()
```

```python
# 点亮 LED0
ol.leds_gpio.write(0, 0x1)
time.sleep(1)

# 熄灭
ol.leds_gpio.write(0, 0x0)
```

```python
import time

def test_leds():
    # 测试 leds_gpio
    print("测试 leds_gpio...")
    
    # 通道1：先设为输出，再写值
    ol.leds_gpio.write(0x04, 0x0)  # 三态寄存器设为输出
    for val in [0x1, 0x2, 0x4, 0x8, 0xF, 0x0]:
        ol.leds_gpio.write(0x00, val)
        print(f"  通道1 值={val}")
        time.sleep(0.5)
    
    # 通道2：先设为输出，再写值
    ol.leds_gpio.write(0x0C, 0x0)  # 三态寄存器设为输出
    for val in [0x1, 0x2, 0x4, 0x8, 0xF, 0x0]:
        ol.leds_gpio.write(0x08, val)
        print(f"  通道2 值={val}")
        time.sleep(0.5)
    
    # 测试 rgbleds_gpio
    print("测试 rgbleds_gpio...")
    ol.rgbleds_gpio.write(0x04, 0x0)
    for val in [0x1, 0x2, 0x4, 0x8, 0xF, 0x0]:
        ol.rgbleds_gpio.write(0x00, val)
        print(f"  通道1 值={val}")
        time.sleep(0.5)
    
    # 最后熄灭
    ol.leds_gpio.write(0x00, 0)
    ol.rgbleds_gpio.write(0x00, 0)
    print("测试结束")

test_leds()
```
